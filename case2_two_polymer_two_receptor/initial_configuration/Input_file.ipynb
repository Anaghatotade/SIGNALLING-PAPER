```python
###   N1 = no. of monomer beads in polymer type 1
###   P1 = no. of polymers of type 1
###   N2 = no. of monomer beads in polymer type 2
###   P2 = no. of polymers of type 2

def input_file(P1, P2, N1, N2):
    import numpy as np
    np.random.seed(42)

    # Total atoms and bonds
    total_atoms = P1*N1 + P2*N2
    total_bonds = P1*(N1-1) + P2*(N2-1)

    file_object = open("initial.xyz", "w+")

    file_object.write("\n{:d} atoms \n{:d} bonds".format(total_atoms, total_bonds))
    file_object.write("\n\n2\t atom types\n1\t bond types")

    # Box size: 80 x 80 x 70
    file_object.write("\n\n-40 40 xlo xhi")
    file_object.write("\n-40 40 ylo yhi")
    file_object.write("\n-35 35 zlo zhi")

    file_object.write("\n\nMasses \n\n1\t1.0 \n2\t1.0 \n")
    file_object.write("\n\nAtoms \n \n")

    particle_index = 1
    bond_index = 1
    molecule_id = 1


    # Polymer Type 1

    for poly in range(P1):
        x_0 = -35 + np.random.random()*35
        y_0 = -35 + np.random.random()*35
        z_0 = -30 + np.random.random()*30

        for bead in range(N1):
            file_object.write("\t%d\t%d\t%d\t%0.2f\t%0.2f\t%0.2f\n" %
                              (particle_index, molecule_id, 1,
                               x_0 + 0.5*bead, y_0, z_0))
            particle_index += 1

        molecule_id += 1


    # Polymer Type 2

    for poly in range(P2):
        x_0 = 0 + np.random.random()*35
        y_0 = 0 + np.random.random()*35
        z_0 = 0 + np.random.random()*30

        for bead in range(N2):
            file_object.write("\t%d\t%d\t%d\t%0.2f\t%0.2f\t%0.2f\n" %
                              (particle_index, molecule_id, 2,
                               x_0 + 0.5*bead, y_0, z_0))
            particle_index += 1

        molecule_id += 1

    # Bonds Section

    file_object.write("\nBonds \n \n")

    atom_counter = 1

    # Bonds for Type 1
    for poly in range(P1):
        for b in range(N1 - 1):
            file_object.write("\t%d\t%d\t%d\t%d\n" %
                              (bond_index, 1,
                               atom_counter + b,
                               atom_counter + b + 1))
            bond_index += 1
        atom_counter += N1

    # Bonds for Type 2
    for poly in range(P2):
        for b in range(N2 - 1):
            file_object.write("\t%d\t%d\t%d\t%d\n" %
                              (bond_index, 1,
                               atom_counter + b,
                               atom_counter + b + 1))
            bond_index += 1
        atom_counter += N2

    file_object.close()
```
