```python
###   N = no. of monomer beads in a polymer
###   P = no. of polymers
def input_file(P1, P2 ,N1,N2):
    import numpy as np
    np.random.seed(42)
    file_object=open("initial.xyz","w+")
    file_object.write("\n{:d} atoms \n{:d} bonds".format(N*P,(N-1)*P))
    file_object.write("\n\n2\t atom types\n1\t bond types")
    file_object.write("\n\n-30 30 xlo xhi\n-30 30 ylo yhi\n-35 35 zlo zhi")
    file_object.write("\n\nMasses \n\n1\t1.0 \n2\t1.0 \n")
    file_object.write("\n\nAtoms \n \n")

    particle_index, bond_index=1,1
    
    ##Put down monomer beads
    for poly_wa in range(P):
        y_0=-20+np.random.random()*40
        x_0=-20+np.random.random()*40
        z_0=-20+np.random.random()*40
        file_object.write("\t%d\t%d\t%d\t%0.2f\t%0.2f\t%0.2f\n" %(1+poly_wa*N,poly_wa+1,1,x_0,y_0,z_0))
        
        for p_wa in np.arange(1,N):
            file_object.write("\t%d\t%d\t%d\t%0.2f\t%0.2f\t%0.2f\n" %(1+p_wa+poly_wa*N,poly_wa+1,1,x_0+0.25*p_wa,y_0,z_0))
            
    
    file_object.write("\nBonds \n \n")
    
    for poly_wa in range(P):
        for bond_wa in np.arange(1,N):
            file_object.write("\t%d\t%d\t%d\t%d\n" %(bond_index,1,poly_wa*N+bond_wa,poly_wa*N+bond_wa+1))
            bond_index+=1

    file_object.close()
```
